```python
orders = orders.drop_duplicates()
order_items = order_items.drop_duplicates()
products = products.drop_duplicates()
customers = customers.drop_duplicates()
suppliers = suppliers.drop_duplicates()
shipments = shipments.drop_duplicates()
inventory = inventory.drop_duplicates()
```

```python
orders["region"] = orders["region"].str.strip()
orders["warehouse"] = orders["warehouse"].str.strip()
orders["sales_channel"] = orders["sales_channel"].str.strip()
orders["order_status"] = orders["order_status"].str.strip()

products["product_name"] = products["product_name"].str.strip()
products["category"] = products["category"].str.strip()

customers["customer_segment"] = customers["customer_segment"].str.strip()
customers["customer_region"] = customers["customer_region"].str.strip()
```

```python
orders["order_date"] = pd.to_datetime(orders["order_date"])

shipments["promised_delivery_date"] = pd.to_datetime(shipments["promised_delivery_date"])

shipments["actual_delivery_date"] = pd.to_datetime(shipments["actual_delivery_date"])
```

```python
order_items["sales"] = (
    order_items["quantity"] *
    order_items["unit_price"] *
    (1 - order_items["discount_pct"])
)

print("Total Sales:", order_items["sales"].sum())
```

```python
order_items["cost"] = (
    order_items["quantity"] * order_items["unit_cost"]
)

order_items["profit"] = (
    order_items["sales"] - order_items["cost"]
)

print("Total Profit:", order_items["profit"].sum())
```

```python
result = pd.merge(order_items, products, on="product_id")

category_sales = result.groupby("category")["sales"].sum()

print(category_sales)
```

```python
region_orders = orders["region"].value_counts()

plt.bar(region_orders.index, region_orders.values)

plt.title("Orders by Region")
plt.xlabel("Region")
plt.ylabel("Number of Orders")

plt.show()
```

```python
status_orders = orders["order_status"].value_counts()

plt.pie(
    status_orders.values,
    labels=status_orders.index,
    autopct="%1.1f%%"
)

plt.title("Orders by Status")

plt.show()
```
